```python
"""
Exercise 15.1 – Steady-state effects of a change in the corporate income tax rate

Usage (from terminal / Anaconda prompt):
    conda activate ogcore-dev
    python exercise15_1_cit_reform.py
"""

import time
import ogcore


def run_baseline():
    print("Running BASELINE steady state...")

    # Default parameters, baseline = True
    p_base = ogcore.parameters.Specifications(
        output_base="./OUTPUT_BASELINE",
        baseline=True,
    )

    t0 = time.time()
    # Steady state only: time_path=False
    ogcore.execute.runner(p_base, time_path=False)
    t1 = time.time()

    print(f"Baseline SS finished in {(t1 - t0)/60:.2f} minutes.\n")


def run_reform(cit_rate_new=0.15):
    print(f"Running REFORM steady state with cit_rate = {cit_rate_new:.2f}...")

    # Start from default params but mark as non-baseline and
    # send output to a different folder
    p_ref = ogcore.parameters.Specifications(
        output_base="./OUTPUT_REFORM",
        baseline=False,
    )

    # OG-Core expects corporate income tax as a list of lists
    # (over time × industries), so even a scalar must be [[x]]
    p_ref.update_specifications({"cit_rate": [[cit_rate_new]]})

    t0 = time.time()
    ogcore.execute.runner(p_ref, time_path=False)
    t1 = time.time()

    print(f"Reform SS finished in {(t1 - t0)/60:.2f} minutes.\n")


if __name__ == "__main__":
    # 1. Baseline SS with default 21% corporate income tax
    run_baseline()

    # 2. Reform SS with new corporate income tax rate (e.g., 15%)
    run_reform(cit_rate_new=0.15)

    print("Done. Compare ./OUTPUT_BASELINE and ./OUTPUT_REFORM for effects.")
```

```python
"""
Exercise 16.1 – Effect of changing beta on steady-state savings profiles

Usage (from terminal / Anaconda prompt):

    conda activate ogcore-dev
    python exercise16_1_beta_profiles.py
"""

import os
import time
import pickle

import ogcore
from ogcore.utils import safe_read_pickle
import ogcore.output_plots as op
import matplotlib.pyplot as plt

# Use separate folders so you don't overwrite other experiments
BASE_OUT = "./OUTPUT_BETA_BASELINE"
REF_OUT  = "./OUTPUT_BETA_REFORM"


def run_baseline():
    print("Running BASELINE SS with default beta...")

    p_base = ogcore.parameters.Specifications(
        baseline=True,
        output_base=BASE_OUT,
        baseline_dir=BASE_OUT,
    )

    t0 = time.time()
    ogcore.execute.runner(p_base, time_path=False)  # SS only
    t1 = time.time()

    print(f"Baseline SS finished in {(t1 - t0)/60:.2f} minutes.\n")
    return p_base


def run_reform():
    print("Running REFORM SS with beta = 0.90...")

    p_ref = ogcore.parameters.Specifications(
        baseline=False,
        output_base=REF_OUT,
        baseline_dir=BASE_OUT,
    )

    # beta is a vector over ability types (shape J)
    import numpy as np
    p_ref.beta = np.full_like(p_ref.beta, 0.90)

    t0 = time.time()
    ogcore.execute.runner(p_ref, time_path=False)   # SS only
    t1 = time.time()

    print(f"Reform SS (beta=0.90) finished in {(t1 - t0)/60:.2f} minutes.\n")
    return p_ref


def load_ss_outputs():
    """Load steady-state dictionaries for baseline and reform."""
    ss_base = safe_read_pickle(os.path.join(BASE_OUT, "SS", "SS_vars.pkl"))
    ss_ref  = safe_read_pickle(os.path.join(REF_OUT,  "SS", "SS_vars.pkl"))
    return ss_base, ss_ref


def plot_savings_profiles(ss_base, ss_ref, p_base, p_ref):
    """
    Plot savings-by-age profiles in baseline vs. beta=0.90 reform
    """
    fig = op.ss_profiles(
        base_ss=ss_base,
        base_params=p_base,
        reform_ss=ss_ref,
        reform_params=p_ref,
        by_j=False,                 # average across ability types
        var="b_sp1",                # savings one period ahead
        plot_title="Savings by age: baseline (β=0.96) vs reform (β=0.90)",
        path=None                   # ← ← DO NOT SAVE
    )
    plt.show()

    print("Displayed savings profile plot.")
    return fig


if __name__ == "__main__":
    # 1. Baseline SS
    p_base = run_baseline()

    # 2. Reform SS with beta = 0.90
    p_ref = run_reform()

    # 3. Load SS outputs
    ss_base, ss_ref = load_ss_outputs()

    # 4. Plot savings profiles by age
    _ = plot_savings_profiles(ss_base, ss_ref, p_base, p_ref)

    print("Done. Plot displayed for Exercise 16.1.")
```

```python
"""
Exercise 16.2 – Convert OG-Core GDP series to constant local currency units.

Run with:
    conda activate ogcore-dev
    python exercise16_2_gdp_to_local_currency.py
"""

import numpy as np
import pickle
import requests
from io import BytesIO
import pandas as pd


# ---------- Helper to load pickle from GitHub raw URL ----------
def load_pickle(url: str):
    r = requests.get(url)
    r.raise_for_status()
    return pickle.load(BytesIO(r.content))


def main():
    # URLs from the exercise (using raw file endpoints)
    TPI_BASE_URL = (
        "https://github.com/PSLmodels/OG-Core/raw/master/"
        "tests/test_io_data/TPI_vars_baseline.pkl"
    )
    SS_BASE_URL = (
        "https://github.com/PSLmodels/OG-Core/raw/master/"
        "tests/test_io_data/SS_vars_baseline.pkl"
    )
    PARAMS_BASE_URL = (
        "https://github.com/PSLmodels/OG-Core/raw/master/"
        "tests/test_io_data/model_params_baseline.pkl"
    )

    print("Downloading baseline TPI, SS, and parameter files...")
    TPI_base = load_pickle(TPI_BASE_URL)
    SS_base = load_pickle(SS_BASE_URL)
    params_base = load_pickle(PARAMS_BASE_URL)

    # ---------- 1. Extract stationarized GDP series ----------
    # TPI_vars["Y"] is GDP in model (stationarized) units
    Y_stationary = TPI_base["Y"]  # shape: (T,)
    T = params_base.T

    # ---------- 2. Add back trend (population + productivity growth) ----------
    # g_n can be an array over T, g_y is long-run productivity growth
    g_n = np.array(params_base.g_n).reshape(-1)
    # If g_n shorter than T, we tile it; if longer, we truncate
    if g_n.size < T:
        g_n = np.resize(g_n, T)
    elif g_n.size > T:
        g_n = g_n[:T]

    g_y = params_base.g_y  # scalar

    # cumulative population growth
    pop_growth = np.cumprod(1 + g_n)
    # productivity growth term exp(g_y * t)
    prod_growth = np.exp(g_y * np.arange(T))

    # Non-stationary GDP in "model units"
    Y_nonstationary = Y_stationary * pop_growth * prod_growth

    # ---------- 3. Convert to real local currency using factor_ss ----------
    factor_ss = SS_base["factor_ss"]
    Y_local_currency = Y_nonstationary * factor_ss

    # ---------- 4. Print and save a small summary ----------
    df = pd.DataFrame(
        {
            "t": np.arange(T),
            "Y_stationary": Y_stationary,
            "Y_nonstationary": Y_nonstationary,
            "Y_local_currency": Y_local_currency,
        }
    )

    print("\nFirst 10 observations of GDP in local currency:")
    print(df.head(10))

    out_csv = "exercise16_2_gdp_local_currency.csv"
    df.to_csv(out_csv, index=False)
    print(f"\nSaved full series to {out_csv}")


if __name__ == "__main__":
    main()
```

```python
"""
Exercise 17.4 – Population distribution over age via get_pop_objs()

Usage:
    conda activate ogcore-dev
    python exercise17_4_pop_by_age.py
"""
import numpy as np
import matplotlib.pyplot as plt
import ogcore.demographics as dem


def main():
    # ------------------------------------------------------------------
    # Parameters for the exercise
    # ------------------------------------------------------------------
    E = 20          # non-economic periods
    S = 80          # working/retired periods
    T = 320         # TPI horizon
    CURR_YEAR = 2023
    FINAL_DATA_YEAR = 2026  # short window for illustration

    # ------------------------------------------------------------------
    # Get population objects from OG-Core
    # ------------------------------------------------------------------
    pop_dict = dem.get_pop_objs(
        E=E,
        S=S,
        T=T,
        initial_data_year=CURR_YEAR,
        final_data_year=FINAL_DATA_YEAR,
        # leave country_id at default '840' (US) unless you want to change it
        GraphDiag=True,        # prints the diagnostics you saw
        download_path=None,
    )

    print("Keys in pop_dict:", pop_dict.keys())

    # -------- Exercise 17.4: population distributions over time --------
    # Training notes say the distribution is stored under key 'omega'
    if "omega" in pop_dict:
        omega = pop_dict["omega"]          # shape: (T, S) in the notes
    else:
        # Fallback for versions that use 'omega_path_S' (T+S, S)
        omega = pop_dict["omega_path_S"]

    print("omega shape:", omega.shape)

    # Age index (0..S-1 or you can shift by E if you want “real” ages)
    ages = np.arange(S)

    # Choose some time indices: 1st, 20th, 100th, last
    # (convert to zero-based indices)
    t_indices = [0, 19, 99, omega.shape[0] - 1]
    t_labels = [
        "t = 1",
        "t = 20",
        "t = 100",
        f"t = {omega.shape[0]} (last)",
    ]

    plt.figure()
    for t, label in zip(t_indices, t_labels):
        if t < omega.shape[0]:
            plt.plot(ages, omega[t, :], label=label)

    plt.xlabel("Model age index")
    plt.ylabel("Population share")
    plt.title("Population distribution over time (omega)")
    plt.legend()
    plt.tight_layout()
    plt.show()

    # -------- Exercise 17.5: population growth rate g_n --------
    g_n = pop_dict["g_n"]     # time-path of population growth rates

    plt.figure()
    plt.plot(np.arange(len(g_n)), g_n)
    plt.xlabel("Time (model periods)")
    plt.ylabel("Population growth rate g_n")
    plt.title("Time path of population growth (g_n)")
    plt.axhline(0, linestyle="--")
    plt.tight_layout()
    plt.show()


if __name__ == "__main__":
    main()
```

```python
"""
Exercise 17.6 – Demographic reform in OG-Core (steady state only)

Usage:
    conda activate ogcore-dev
    python exercise17_6_demog_reform.py
"""

import time
import os
import pickle

import ogcore
import ogcore.demographics as dem
import ogcore.output_tables as ot


# -------------------------------------------------------------------
# Helper: run baseline SS with default demographics
# -------------------------------------------------------------------
def run_baseline():
    print("=== Running BASELINE SS with default demographics ===")

    BASE_OUT = "./OUTPUT_DEMOG_BASELINE"

    p_base = ogcore.parameters.Specifications(
        baseline=True,
        baseline_dir=BASE_OUT,
        output_base=BASE_OUT,
    )

    t0 = time.time()
    ogcore.execute.runner(p_base, time_path=False)  # SS only
    t1 = time.time()

    print(f"Baseline SS finished in {(t1 - t0)/60:.2f} minutes.\n")
    return p_base


# -------------------------------------------------------------------
# Helper: run reform SS with updated demographics from get_pop_objs
# -------------------------------------------------------------------
def run_demog_reform():
    print("=== Running DEMOGRAPHIC REFORM SS ===")

    E, S, T = 20, 80, 320
    COUNTRY_ID = "840"
    INITIAL_DATA_YEAR = 2023
    FINAL_DATA_YEAR   = 2026

    BASE_OUT = "./OUTPUT_DEMOG_BASELINE"
    REF_OUT  = "./OUTPUT_DEMOG_REFORM"

    # Build pop objects (omega_path_S, g_n_path, etc.)
    pop_dict = dem.get_pop_objs(
        E=E,
        S=S,
        T=T,
        country_id=COUNTRY_ID,
        initial_data_year=INITIAL_DATA_YEAR,
        final_data_year=FINAL_DATA_YEAR,
        GraphDiag=False,
        download_path=None,
    )

    # Start with default specs, mark as non-baseline but
    # point to the baseline directory (for TR, r guesses, etc.)
    p_ref = ogcore.parameters.Specifications(
        baseline=False,
        baseline_dir=BASE_OUT,
        output_base=REF_OUT,
    )

    # Update demographic parameters in the Specifications object
    # pop_dict keys match the names used in OG-Core (e.g. omega_path_S, g_n_SS, etc.)
    p_ref.update_specifications(pop_dict)

    t0 = time.time()
    ogcore.execute.runner(p_ref, time_path=False)  # SS only
    t1 = time.time()

    print(f"Demographic reform SS finished in {(t1 - t0)/60:.2f} minutes.\n")
    return p_ref


# -------------------------------------------------------------------
# Helper: load SS output dictionaries
# -------------------------------------------------------------------
def load_ss_output(output_base):
    ss_path = os.path.join(output_base, "SS", "SS_vars.pkl")
    with open(ss_path, "rb") as f:
        ss_dict = pickle.load(f)
    return ss_dict


def main():
    # 1. Baseline steady state
    p_base = run_baseline()

    # 2. Demographic reform steady state
    p_ref = run_demog_reform()

    # 3. Load SS outputs from disk
    ss_base = load_ss_output("./OUTPUT_DEMOG_BASELINE")
    ss_reform = load_ss_output("./OUTPUT_DEMOG_REFORM")

    # 4. Build macro comparison table (steady-state only)
    # macro_table_SS compares SS aggregates between two runs
    table = ot.macro_table_SS(
        ss_base,
        ss_reform,
        table_format="pct_diff"   # <-- this is the correct keyword
    )

    print("\n=== Steady-state macro variables: reform vs baseline (% diff) ===")
    # if to_markdown is available:
    try:
        print(table.to_markdown(index=True))
    except AttributeError:
        # fallback if pandas is older
        print(table.to_string(index=True))

    # For your write-up: comment on how GDP, K, L, wages, r, etc. change.

if __name__ == "__main__":
    main()
```

```python
# --------------------------------------------------------------
# Exercise 17.10: Estimate AR(1) process for GDP growth
# g_t = α + ρ g_{t-1} + ε_t
# --------------------------------------------------------------
import statsmodels.api as sm

# We use the quarterly GDP data from Exercise 17.7
gdp_q = web.DataReader(FRED_SERIES_ID, "fred", START_DATE, END_DATE)
gdp_q.rename(columns={FRED_SERIES_ID: "RGDP"}, inplace=True)

# Quarterly GDP growth (%)
gdp_q["growth"] = gdp_q["RGDP"].pct_change() * 100
gdp_q = gdp_q.dropna()

# Lagged growth
gdp_q["growth_lag"] = gdp_q["growth"].shift(1)
gdp_q = gdp_q.dropna()

# Prepare regression variables
Y = gdp_q["growth"]
X = sm.add_constant(gdp_q["growth_lag"])

# Run OLS: AR(1)
model = sm.OLS(Y, X)
results = model.fit()

print("\n===== AR(1) Estimation Results for South Africa GDP Growth =====")
print(results.summary())

alpha_hat = results.params["const"]
rho_hat   = results.params["growth_lag"]

print("\nEstimated parameters:")
print(f"α (constant) = {alpha_hat:.4f}")
print(f"ρ (rho)      = {rho_hat:.4f}")

# Interpretation
if abs(rho_hat) < 0.3:
    persistence = "low"
elif abs(rho_hat) < 0.7:
    persistence = "moderate"
else:
    persistence = "high"

print("\nInterpretation:")
print(f"- The AR(1) coefficient ρ = {rho_hat:.3f} indicates **{persistence} persistence**.")
print("- Positive ρ means good quarters tend to follow good quarters (momentum).")
print("- If ρ is small, shocks do not last long in South Africa’s growth process.")
```

```python
"""
Exercise 17.10 – AR(1) process for South African GDP growth

This script:
1. Downloads quarterly real GDP for South Africa from FRED.
2. Computes quarterly GDP growth (log-difference * 100).
3. Estimates an AR(1) model:
       y_t = alpha + rho * y_{t-1} + eps_t
   using statsmodels.OLS.
4. Prints the regression summary and the estimated alpha, rho.
"""

import pandas as pd
import numpy as np
from pandas_datareader import data as web
import statsmodels.api as sm


def build_gdp_growth_series():
    """
    Download South African real GDP (quarterly) from FRED
    and construct a growth series.

    Returns
    -------
    gdp_growth : pd.DataFrame
        Columns: ["date", "growth"]
        where growth is 100 * log(RGDP_t / RGDP_{t-1}).
    """
    # FRED series ID used in the training material:
    series_id = "NGDPRSAXDCZAQ"  # Real GDP, South Africa, SA, quarterly

    # Download from FRED
    gdp = web.DataReader(series_id, "fred", start="1994-01-01")
    gdp = gdp.dropna()
    gdp = gdp.rename(columns={series_id: "RGDP"})

    # Log growth * 100 (approximate percent growth)
    gdp["growth"] = 100.0 * (
        np.log(gdp["RGDP"]) - np.log(gdp["RGDP"].shift(1))
    )

    # Drop first NaN from lag and reset index
    gdp = gdp.dropna(subset=["growth"]).reset_index()
    gdp = gdp.rename(columns={"DATE": "date"})

    # Return only what we need
    return gdp[["date", "growth"]]


def estimate_ar1(gdp_growth: pd.DataFrame):
    """
    Estimate AR(1): y_t = alpha + rho * y_{t-1} + eps_t

    Parameters
    ----------
    gdp_growth : DataFrame
        Must have a column "growth".
    """
    df = gdp_growth.copy()
    df = df.rename(columns={"growth": "y"})
    df["y_lag"] = df["y"].shift(1)
    df = df.dropna()

    # Design matrix: constant + lagged y
    X = sm.add_constant(df["y_lag"])
    y = df["y"]

    model = sm.OLS(y, X).fit()

    print(model.summary())

    alpha_hat = model.params["const"]
    rho_hat = model.params["y_lag"]

    print("\n=== AR(1) estimates for South African GDP growth ===")
    print(f"alpha (intercept) = {alpha_hat:.4f}")
    print(f"rho (AR(1) coefficient) = {rho_hat:.4f}")

    # Quick interpretation text you can reuse in your writeup
    print("\nInterpretation for Exercise 17.10 (you can paraphrase this):")
    print(
        f"- The constant alpha ≈ {alpha_hat:.2f} is the mean quarterly growth "
        "rate implied by the AR(1) process (once shocks die out)."
    )
    print(
        f"- The AR(1) coefficient rho ≈ {rho_hat:.2f} measures persistence: "
        "if it's close to 1, shocks to growth are highly persistent; if closer "
        "to 0, growth reverts quickly to its mean."
    )


def main():
    # 1) Build growth series
    gdp_growth = build_gdp_growth_series()
    print("First few observations of GDP growth:")
    print(gdp_growth.head(), "\n")

    # 2) Estimate AR(1)
    estimate_ar1(gdp_growth)


if __name__ == "__main__":
    main()
```

```python
"""
Exercise 17.12 – Adjust earnings process to match a target-country Gini

This script:
  1. Loads the baseline U.S. earnings process from ogcore.parameters.Specifications().
  2. Extracts an S×J earnings matrix (age × ability) from p.e
     (handling the case where p.e is 2D or 3D).
  3. Computes the U.S. Gini from that matrix.
  4. Scales dispersion via e_c = mean(e_US) + phi * (e_US - mean(e_US)),
     and chooses phi to match a target-country Gini using scipy.optimize.
  5. Plots U.S. vs target-country earnings profiles by age for each ability type.
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize_scalar
import ogcore

# =======================================================
# 0. Load earnings matrix from OG-Core
# =======================================================

# Load default parameter object
p = ogcore.parameters.Specifications()

# p.e may be 2D (S×J) or 3D (T×S×J) depending on OG-Core version
e_raw = p.e
print("Raw p.e shape:", e_raw.shape)

if e_raw.ndim == 2:
    # Already S×J
    e_US = e_raw
    S, J = e_US.shape
elif e_raw.ndim == 3:
    # Likely T×S×J. Take the last time period as a representative cross-section.
    T, S, J = e_raw.shape
    e_US = e_raw[-1, :, :]   # shape S×J
else:
    raise RuntimeError(
        f"Unexpected dimensionality for p.e: ndim={e_raw.ndim}, "
        f"shape={e_raw.shape}. Expected 2D or 3D."
    )

print(f"Using earnings matrix of shape S={S}, J={J}")
print("First few entries:\n", e_US[:5, :min(5, J)])


# =======================================================
# 1. User inputs: Gini targets
# =======================================================

# Gini indices in 0–100 units
GINI_US_2019 = 41.5           # reference value in the text
GINI_TARGET_COUNTRY = 63.0    # example: South Africa ≈ 63
TARGET_COUNTRY_NAME = "South Africa"


# =======================================================
# 2. Gini helper
# =======================================================

def gini_from_matrix(e_mat: np.ndarray) -> float:
    """
    Compute the Gini coefficient (0–100) for an earnings matrix e_mat (S×J),
    treating each cell as equally weighted.

    You can later extend this to use population weights by age/ability.
    """
    y = e_mat.ravel().astype(float)

    # Ensure non-negative (if any log/negative values slipped in)
    if np.any(y < 0):
        y = y - np.min(y) + 1e-8

    if np.all(y == 0):
        return 0.0

    y_sorted = np.sort(y)
    n = len(y_sorted)
    cum_y = np.cumsum(y_sorted)
    total_y = cum_y[-1]

    i = np.arange(1, n + 1)
    gini = 1.0 - 2.0 * np.sum((n + 1 - i) * y_sorted) / (n * total_y)

    return 100.0 * gini  # scale to 0–100


# =======================================================
# 3. Transformation: scale dispersion with phi
# =======================================================

def transform_earnings(e_us: np.ndarray, phi: float) -> np.ndarray:
    """
    Keep mean earnings fixed but rescale dispersion via phi:

        e_c = mean(e_us) + phi * (e_us - mean(e_us))

    - phi > 1 → more dispersion (higher inequality)
    - phi < 1 → less dispersion (lower inequality)
    """
    mean_e = e_us.mean()
    return mean_e + phi * (e_us - mean_e)


# =======================================================
# 4. Objective function for matching target Gini
# =======================================================

def gini_gap_objective(phi: float, e_us: np.ndarray, gini_target: float) -> float:
    """
    For a given phi, compute the squared difference between
    Gini(e_c(phi)) and gini_target.
    """
    e_c = transform_earnings(e_us, phi)
    gini_c = gini_from_matrix(e_c)
    return (gini_c - gini_target) ** 2


# =======================================================
# 5. Main calibration and plotting
# =======================================================

def main():
    # Baseline Gini from e_US
    gini_us_empirical = gini_from_matrix(e_US)
    print(f"\nComputed U.S. Gini from e_US: {gini_us_empirical:.2f} "
          f"(vs. stated {GINI_US_2019:.1f})")

    print(f"Target-country Gini ({TARGET_COUNTRY_NAME}): {GINI_TARGET_COUNTRY:.2f}")

    # Calibrate phi
    res = minimize_scalar(
        gini_gap_objective,
        bounds=(0.1, 5.0),
        args=(e_US, GINI_TARGET_COUNTRY),
        method="bounded",
        options={"xatol": 1e-4}
    )

    phi_star = res.x
    print(f"\nCalibrated phi for {TARGET_COUNTRY_NAME}: {phi_star:.4f}")
    print(f"Objective value (squared Gini gap): {res.fun:.6f}")

    # Build transformed earnings matrix
    e_country = transform_earnings(e_US, phi_star)
    gini_country_empirical = gini_from_matrix(e_country)
    print(f"Gini for transformed earnings in {TARGET_COUNTRY_NAME}: "
          f"{gini_country_empirical:.2f}")

    # Plot by age for each ability type
    ages = np.arange(S)

    plt.figure(figsize=(8, 5))
    for j in range(J):
        # U.S. profile
        plt.plot(
            ages,
            e_US[:, j],
            linestyle="--",
            alpha=0.6,
            label=f"U.S., type {j+1}" if j < 3 else None  # limit legend clutter
        )
        # Target country profile
        plt.plot(
            ages,
            e_country[:, j],
            alpha=0.9,
            label=f"{TARGET_COUNTRY_NAME}, type {j+1}" if j < 3 else None
        )

    plt.xlabel("Age index")
    plt.ylabel("Earnings (model units)")
    plt.title(
        f"Earnings profiles: U.S. vs. {TARGET_COUNTRY_NAME}\n"
        f"Gini_US ≈ {gini_us_empirical:.1f}, "
        f"Gini_{TARGET_COUNTRY_NAME} ≈ {gini_country_empirical:.1f}"
    )
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()

    print("\nInterpretation hint:")
    print("- If phi* > 1, the transformed country has more dispersion and higher Gini than the U.S.")
    print("- If phi* < 1, earnings profiles are compressed around the mean, lowering Gini.")
    print("- Visually, the higher-Gini country should show larger gaps between ability types at each age.")


if __name__ == "__main__":
    main()
```
